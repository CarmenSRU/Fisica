### Qué mirar

- Con velocidades bajas (ej. 0.05c y 0.1c) las dos fórmulas casi coinciden — así se comportaba el mundo para Galileo.
- Sube el tren a 0.9c o más: la diferencia entre **Galileana** y **Especial** se vuelve grande, y la relativista **nunca supera c**.
- Con ambas velocidades cerca de 0.99c, la Galileana da una velocidad relativa casi nula, pero la Especial se mantiene siempre por debajo de c — mira cómo cambia la fórmula ya reemplazada en cada caso.


## Simulación 2 — Dilatación del tiempo: el viaje a Marte

Dos gemelos de la misma edad. Uno se queda en la Tierra 🌍, el otro viaja a Marte en una nave 🚀 a una velocidad que tú eliges, durante el tiempo que tú elijas (medido en años, según el reloj de la Tierra).

Arriba de los controles hay dos **relojes de luz** — la idea clásica de un fotón (el puntico dorado) rebotando entre dos espejos. Es *el mecanismo* detrás de la dilatación: para un reloj que se mueve, el fotón tiene que recorrer un camino más largo visto desde la Tierra, así que "tic-tac" más lento. Por eso el reloj del viajero rebota más despacio mientras más rápido va la nave — y por eso, al llegar, el viajero es más joven. La animación también avanza más rápido o más lento según la velocidad que elijas.

Dentro del panel, debajo de los controles, está la fórmula simbólica y, justo abajo, la misma fórmula ya reemplazada con tus valores actuales, actualizándose en vivo. El número de "EN LA TIERRA" es siempre el mismo que pusiste en el control de duración — por definición, es el tiempo que pasa en la Tierra durante todo el viaje.

*(Nota: es una versión simplificada de velocidad constante — el "problema de los gemelos" real necesita que la nave frene y regrese, lo cual trae aceleración a la historia. Aquí nos quedamos con la idea central: el tiempo propio del viajero es menor.)*

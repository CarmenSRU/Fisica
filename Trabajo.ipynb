# Introducción Interactiva a la Teoría de la Relatividad Especial de Einstein frente a la Mecánica Clásica (Galileana)

Este notebook desarrolla una introducción interactiva a la *Teoría de la Relatividad Especial de Einstein* en comparación con la *Mecánica Clásica (Galileana)*. La explicación teórica se estructura en tres secciones fundamentales:

---

## 1. Adición de Velocidades: Galileana vs. Especial

En la física clásica (Galileo/Newton), si un cuerpo se mueve a velocidad $v_1$ respecto a un medio y el medio a $v_2$, las velocidades simplemente se suman o restan. En la relatividad especial, la velocidad de la luz $c$ es un límite insuperable y una constante en todos los sistemas de referencia inerciales.

### Transformación Galileana (Adición clásica)

$$v_{\text{rel}} = v_{\text{tren}} - v_{\text{ciclista}}$$

### Transformación Relativista (Einstein)

$$v_{\text{rel}} = \frac{v_{\text{tren}} - v_{\text{ciclista}}}{1 - \frac{v_{\text{tren}} \cdot v_{\text{ciclista}}}{c^2}}$$

* *Comportamiento a bajas velocidades ($v \ll c$):* El término $\frac{v_{\text{tren}} \cdot v_{\text{ciclista}}}{c^2}$ es prácticamente cero, por lo que la fórmula relativista equivale a la galileana.
* *Comportamiento a altas velocidades ($v \to c$):* El denominador aumenta, garantizando que la velocidad relativa resultante nunca supere la velocidad de la luz $c$.

---

## 2. Dilatación del Tiempo (El Viaje a Marte y la Paradoja de los Gemelos)

El tiempo no es absoluto; transcurre más lento para un objeto o viajero en movimiento en comparación con un observador en reposo.

### Factor de Lorentz ($\gamma$)

$$\gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}$$

### Fórmula de Dilatación Temporal

$$\Delta t_{\text{viajero}} = \frac{\Delta t_{\text{Tierra}}}{\gamma} = \Delta t_{\text{Tierra}} \cdot \sqrt{1 - \frac{v^2}{c^2}}$$

* *Relojes de luz:* Un pulso de luz rebotando entre dos espejos recorre una distancia mayor (en diagonal) para un observador externo que ve el reloj en movimiento. Dado que la velocidad de la luz $c$ debe mantenerse constante, el intervalo de tiempo entre cada "tic-tac" debe ser mayor.
* *Resultado:* Cuanto más cerca se viaja de $c$, mayor es el factor $\gamma$, haciendo que el tiempo para el viajero transcurra más despacio respecto al observador en la Tierra.

---

## 3. Contracción de la Longitud (Contracción de Lorentz)

Los objetos que se desplazan a velocidades cercanas a la de la luz sufren una reducción en la medida de su longitud a lo largo de la dirección de su movimiento, vista por un observador externo en reposo.

### Fórmula de Contracción de Longitud

$$L = \frac{L_0}{\gamma} = L_0 \cdot \sqrt{1 - \frac{v^2}{c^2}}$$

* *Longitud propia ($L_0$):* La longitud del objeto medida por un observador en reposo respecto al propio objeto.
* *Longitud observada ($L$):* La longitud contraída que mide el observador externo.
* *Efecto visual:* A $v \approx 0$, $L \approx L_0$. Conforme $v \to c$, la longitud del objeto en la dirección del movimiento tiende a cero.

```python
from IPython.display import HTML

HTML(r'''
<div class="sim1-root" id="sim1-root">
  <style>
    .sim1-root{
      --bg-deep:#0B1023; --bg-mid:#141B3D; --line:#2A3363;
      --text:#F2F4FA; --text-dim:#8B93BE;
      --gold:#F2B705; --violet:#9D7BEA; --teal:#4FD1C5;
      font-family:-apple-system,BlinkMacSystemFont,"Segoe UI",Roboto,sans-serif;
      color:var(--text); max-width:820px; width:100%; margin:0 auto;
      border-radius:18px; overflow:hidden;
      background:linear-gradient(180deg,var(--bg-deep),var(--bg-mid));
      box-shadow:0 20px 50px rgba(0,0,0,.35);
    }
    .sim1-sky{
      position:relative; height:190px; overflow:hidden;
      background:
        radial-gradient(2px 2px at 10% 20%, #fff8 40%, transparent 41%),
        radial-gradient(2px 2px at 25% 60%, #fff6 40%, transparent 41%),
        radial-gradient(1.5px 1.5px at 40% 30%, #fff5 40%, transparent 41%),
        radial-gradient(1.5px 1.5px at 65% 15%, #fff7 40%, transparent 41%),
        radial-gradient(2px 2px at 80% 45%, #fff5 40%, transparent 41%),
        radial-gradient(1.5px 1.5px at 92% 70%, #fff6 40%, transparent 41%),
        radial-gradient(1.5px 1.5px at 55% 80%, #fff4 40%, transparent 41%),
        linear-gradient(180deg, var(--bg-deep) 0%, var(--bg-mid) 100%);
    }
    .sim1-track{ position:absolute; left:5%; right:5%; bottom:34px; height:2px; background:var(--line); }
    .sim1-track::after{ content:""; position:absolute; left:0; right:0; top:-1px; height:2px;
      background:repeating-linear-gradient(90deg, var(--gold) 0 10px, transparent 10px 22px); opacity:.55; }
    .sim1-icon{ position:absolute; bottom:-6px; font-size:30px; left:0; transform:translateX(-50%);
      filter:drop-shadow(0 6px 10px rgba(0,0,0,.4)); }
    .sim1-earth{ position:absolute; left:50%; bottom:22px; font-size:34px; transform:translateX(-50%); }
    .sim1-label{ position:absolute; top:-22px; left:50%; transform:translateX(-50%);
      font-size:11px; color:var(--text-dim); white-space:nowrap; }
    .sim1-panel{ padding:18px 22px 22px; }
    .sim1-row{ display:flex; gap:22px; flex-wrap:wrap; margin-bottom:14px; }
    .sim1-slider{ flex:1 1 220px; }
    .sim1-slider label{ display:flex; justify-content:space-between; font-size:13px; color:var(--text-dim); margin-bottom:6px; }
    .sim1-slider label span.val{ color:var(--text); font-family:ui-monospace,Menlo,Consolas,monospace; }
    .sim1-slider input[type=range]{ width:100%; accent-color:var(--gold); }
    .sim1-dirbtn{ border:1px solid var(--line); background:transparent; color:var(--text-dim);
      font-size:11px; padding:2px 7px; border-radius:7px; cursor:pointer; margin-left:6px; }
    .sim1-dirbtn.neg{ color:var(--gold); border-color:var(--gold); }
    .sim1-modes{ display:flex; gap:8px; align-items:center; }
    .sim1-modebtn{ border:1px solid var(--line); background:transparent; color:var(--text-dim);
      padding:8px 14px; border-radius:999px; font-size:13px; cursor:pointer; transition:all .15s; }
    .sim1-modebtn.active[data-mode="Galileana"]{ background:var(--violet); color:#151024; border-color:var(--violet); }
    .sim1-modebtn.active[data-mode="Especial"]{ background:var(--teal); color:#0B1F1D; border-color:var(--teal); }
    .sim1-actions{ display:flex; gap:8px; margin-left:auto; }
    .sim1-actbtn{ border:1px solid var(--line); background:#ffffff0d; color:var(--text);
      padding:8px 14px; border-radius:10px; font-size:13px; cursor:pointer; }
    .sim1-actbtn:hover{ background:#ffffff1a; }
    .sim1-formula{ text-align:center; font-family:ui-monospace,Menlo,Consolas,monospace; font-size:13px;
      color:var(--text-dim); margin:2px 0 16px; line-height:1.8; }
    .sim1-formula .sim1-f-sym{ font-size:13.5px; }
    .sim1-formula .sim1-f-sub{ color:var(--gold); font-size:14px; }
    .sim1-readout{ display:grid; grid-template-columns:repeat(3,1fr); gap:12px; margin-top:4px; }
    .sim1-card{ background:#ffffff0a; border:1px solid var(--line); border-radius:12px; padding:12px 14px; }
    .sim1-card h4{ margin:0 0 8px; font-size:12px; color:var(--text-dim); font-weight:600; }
    .sim1-card .num{ font-family:ui-monospace,Menlo,Consolas,monospace; font-size:15px; line-height:1.7; }
    .sim1-card .num b{ color:var(--gold); font-weight:600; }
    .sim1-rel .num b{ color:var(--teal); }
    @media (max-width:640px){ .sim1-readout{ grid-template-columns:1fr; } }
  </style>

  <div class="sim1-sky">
  <div class="sim1-track" id="sim1-track">
    <div class="sim1-earth">🌍</div>
    
    <div class="sim1-icon" id="sim1-bike">
      <span style="display: inline-block; transform: scaleX(-1);">🚴</span>
      <div class="sim1-label">ciclista</div>
    </div>
    
    <div class="sim1-icon" id="sim1-train">
      <span style="display: inline-block; transform: scaleX(-1);">🚂</span>
      <div class="sim1-label">tren</div>
    </div>

  </div>
</div>

  <div class="sim1-panel">
    <div class="sim1-row">
      <div class="sim1-slider">
        <label>v ciclista <span><span class="val" id="sim1-vb-val">0.05c →</span>
          <button class="sim1-dirbtn" id="sim1-dir-bike" title="Cambiar sentido">⇄ sentido</button></span></label>
        <input type="range" id="sim1-vb" min="0" max="0.99" step="0.01" value="0.05">
      </div>
      <div class="sim1-slider">
        <label>v tren <span><span class="val" id="sim1-vt-val">0.50c →</span>
          <button class="sim1-dirbtn" id="sim1-dir-train" title="Cambiar sentido">⇄ sentido</button></span></label>
        <input type="range" id="sim1-vt" min="0" max="0.99" step="0.01" value="0.50">
      </div>
    </div>
    <div class="sim1-row" style="align-items:center;">
      <div class="sim1-modes" id="sim1-modes">
        <button class="sim1-modebtn active" data-mode="Galileana">Galileana</button>
        <button class="sim1-modebtn" data-mode="Especial">Especial</button>
      </div>
      <div class="sim1-actions">
        <button class="sim1-actbtn" id="sim1-play">⏸ Pausar</button>
        <button class="sim1-actbtn" id="sim1-reset">↺ Reiniciar</button>
      </div>
    </div>

    <div class="sim1-formula">
      <div class="sim1-f-sym" id="sim1-f-sym">v_rel = v_tren − v_ciclista</div>
      <div class="sim1-f-sub" id="sim1-f-sub">v_rel = 0.50c − 0.05c = 0.4500c</div>
    </div>

    <div class="sim1-readout">
      <div class="sim1-card"><h4>CICLISTA (resp. Tierra)</h4>
        <div class="num">x = <b id="sim1-xb">0.00</b> s·c<br>v = <b id="sim1-vb-out">0.05</b>c</div></div>
      <div class="sim1-card"><h4>TREN (resp. Tierra)</h4>
        <div class="num">x = <b id="sim1-xt">0.00</b> s·c<br>v = <b id="sim1-vt-out">0.50</b>c</div></div>
      <div class="sim1-card sim1-rel"><h4>TREN VISTO DESDE CICLISTA</h4>
        <div class="num">modelo: <b id="sim1-mode-out">Galileana</b><br>v_rel = <b id="sim1-vrel">0.4500</b>c</div></div>
    </div>
  </div>
</div>

<script>
(function(){
  const root = document.getElementById('sim1-root');
  const bikeEl = root.querySelector('#sim1-bike');
  const trainEl = root.querySelector('#sim1-train');
  const trackEl = root.querySelector('#sim1-track');
  const vbSlider = root.querySelector('#sim1-vb');
  const vtSlider = root.querySelector('#sim1-vt');
  const modeBtns = root.querySelectorAll('.sim1-modebtn');
  const playBtn = root.querySelector('#sim1-play');
  const resetBtn = root.querySelector('#sim1-reset');
  const dirBikeBtn = root.querySelector('#sim1-dir-bike');
  const dirTrainBtn = root.querySelector('#sim1-dir-train');

  let dirBike = 1, dirTrain = 1;

  let mode = "Galileana";
  let playing = true;
  let t = 0;
  const T_MAX = 20;
  const SIM_SPEED = 3.2;

  function relVel(vb, vt){
    return mode === "Galileana" ? (vt - vb) : (vt - vb) / (1 - vb*vt);
  }

  function fmt(v){
    return v >= 0 ? v.toFixed(2) : `(${v.toFixed(2)})`;
  }

  function render(){
    const magB = parseFloat(vbSlider.value);
    const magT = parseFloat(vtSlider.value);
    const vb = magB * dirBike;
    const vt = magT * dirTrain;
    const xb = vb * t, xt = vt * t;
    const trackW = trackEl.clientWidth;
    const half = trackW / 2;
    const fb = Math.max(-1, Math.min(xb / T_MAX, 1));
    const ft = Math.max(-1, Math.min(xt / T_MAX, 1));

    bikeEl.style.left = (half + fb*half) + 'px';
    trainEl.style.left = (half + ft*half) + 'px';

    root.querySelector('#sim1-vb-val').textContent = magB.toFixed(2) + 'c ' + (dirBike>0?'→':'←');
    root.querySelector('#sim1-vt-val').textContent = magT.toFixed(2) + 'c ' + (dirTrain>0?'→':'←');
    root.querySelector('#sim1-xb').textContent = xb.toFixed(2);
    root.querySelector('#sim1-xt').textContent = xt.toFixed(2);
    root.querySelector('#sim1-vb-out').textContent = vb.toFixed(2);
    root.querySelector('#sim1-vt-out').textContent = vt.toFixed(2);
    root.querySelector('#sim1-mode-out').textContent = mode;
    const vrel = relVel(vb, vt);
    root.querySelector('#sim1-vrel').textContent = vrel.toFixed(4);

    const symEl = root.querySelector('#sim1-f-sym');
    const subEl = root.querySelector('#sim1-f-sub');
    if (mode === "Galileana"){
      symEl.textContent = "v_rel = v_tren − v_ciclista";
      subEl.textContent = `v_rel = ${fmt(vt)}c − ${fmt(vb)}c = ${vrel.toFixed(4)}c`;
    } else {
      symEl.textContent = "v_rel = (v_tren − v_ciclista) / (1 − v_tren·v_ciclista/c²)";
      const num = vt - vb;
      const den = 1 - vb*vt;
      subEl.textContent = `v_rel = (${fmt(vt)}c − ${fmt(vb)}c) / (1 − ${fmt(vt)}×${fmt(vb)}) = ${fmt(num)}c / ${den.toFixed(4)} = ${vrel.toFixed(4)}c`;
    }
  }

  let last = performance.now();
  function loop(now){
    const dt = (now - last) / 1000; last = now;
    if (playing){
      t += dt * SIM_SPEED;
      if (t > T_MAX) t = 0;
    }
    render();
    requestAnimationFrame(loop);
  }

  vbSlider.addEventListener('input', render);
  vtSlider.addEventListener('input', render);
  modeBtns.forEach(btn => btn.addEventListener('click', () => {
    modeBtns.forEach(b => b.classList.remove('active'));
    btn.classList.add('active');
    mode = btn.dataset.mode;
    render();
  }));
  playBtn.addEventListener('click', () => {
    playing = !playing;
    playBtn.textContent = playing ? '⏸ Pausar' : '▶ Reanudar';
  });
  resetBtn.addEventListener('click', () => { t = 0; render(); });
  dirBikeBtn.addEventListener('click', () => {
    dirBike *= -1;
    dirBikeBtn.classList.toggle('neg', dirBike < 0);
    render();
  });
  dirTrainBtn.addEventListener('click', () => {
    dirTrain *= -1;
    dirTrainBtn.classList.toggle('neg', dirTrain < 0);
    render();
  });

  requestAnimationFrame(loop);
})();
</script>

''')
```

### Qué mirar

- Con velocidades bajas (ej. 0.05c y 0.1c, mismo sentido) las dos fórmulas casi coinciden: así se comportaba el mundo para Galileo.
- El caso que de verdad muestra el efecto relativista: pon tren a 0.9c y usa el botón "⇄ sentido" del ciclista para que vaya a 0.9c en sentido contrario. Ahí la Galileana da una velocidad relativa mayor que la de la luz, lo que es imposible, mientras que la Especial siempre queda debajo de c.
- **Sin importar qué tan rápido vayan los dos (incluso en sentidos opuestos), la fórmula especial nunca deja que la velocidad relativa supere a c — la galileana sí lo permite, y por eso sabemos que falla a estas velocidades.**
- La velocidad del tren respecto al ciclista es la velocidad del tren respecto a la Tierra, menos la velocidad del ciclista respecto a la Tierra.

Arriba de los controles hay dos **relojes de luz** — la idea clásica de un fotón (el puntico dorado) rebotando entre dos espejos. Es *el mecanismo* detrás de la dilatación: para un reloj que se mueve, el fotón tiene que recorrer un camino más largo visto desde la Tierra, así que "tic-tac" más lento. # Conclusiones

### 1. La física clásica tiene un rango de aplicación

Las simulaciones permiten observar que las aproximaciones de la física clásica funcionan adecuadamente cuando las velocidades involucradas son pequeñas frente a la velocidad de la luz. Sin embargo, cuando \(v\) se aproxima a \(c\), es necesario utilizar las expresiones de la relatividad especial.
De esta manera, un modelo que funciona en condiciones cotidianas no necesariamente se extrapola a velocidades relativistas.

---

### 2. La relatividad modifica magnitudes que parecen absolutas

Magnitudes como el **tiempo** y la **longitud** dependen del estado de movimiento del observador, por tanto no son cantidades absolutas en el marco de la relatividad especial. Las representaciones interactivas permiten pasar de estas expresiones matemáticas a una interpretación visual, facilitando la comprensión de cómo cambia la percepción de un mismo fenómeno entre diferentes sistemas de referencia.

---

### 3. Efectos físicos de la relatividad en un sistema
Se pudo observar directamente cómo cambia el comportamiento del sistema a por medio de las simulaciones. Esto permite comprender que los efectos relativistas no son únicamente ecuaciones matemáticas, sino que producen diferencias significativas cuando las velocidades son suficientemente altas.

Por ejemplo, una pequeña modificación en la velocidad puede producir cambios en el factor de Lorentz:

$$
\gamma=
\frac{1}{\sqrt{1-\frac{v^2}{c^2}}}
$$

y, por tanto, modificar simultáneamente los tiempos y las longitudes observadas.

---

### 4. Un modelo físico adecuado es fundamental para interpretar correctamente un sistema

Las simulaciones muestran que utilizar una aproximación fuera de su rango de validez puede conducir a resultados diferentes de los predichos por el modelo relativista. Por esta razón, comprender **cuándo debe utilizarse un modelo clásico y cuándo es necesario utilizar la relatividad especial** es fundamental para interpretar correctamente fenómenos físicos y sus aplicaciones tecnológicas.

Esto resulta especialmente relevante en sistemas biomédicos que incorporan partículas aceleradas, sensores y sistemas de posicionamiento de alta precisión.

## Ejemplo: Radiocirugía Robótica de Alta Energía y Composición Relativista de Velocidades
En los sistemas modernos de radiocirugía robótica se combina un brazo robótico que posiciona la fuente y un acelerador lineal miniaturizado que dispara partículas subatómicas cargadas. 

Si el brazo robótico proyecta la fuente en la misma dirección en la que el acelerador emite las partículas, podríamos pensar que sus velocidades simplemente se suman. Sin embargo, a velocidades cercanas a la de la luz ($c$), la mecánica clásica falla y la **composición relativista de velocidades de Einstein** es indispensable para no errar el blanco tumoral ni calcular mal la dosis.

---

### Planteamiento del Problema

Un sistema de radiocirugía robótica consta de un brazo que se desplaza horizontalmente a una velocidad $v_{\text{robot}} = 0.10c$ respecto a la sala de operaciones (marco del laboratorio). Simultáneamente, el acelerador instalado en el brazo dispara un haz de partículas que transporta un radioisótopo trazador a una velocidad $v_{\text{haz}} = 0.85c$ respecto al propio brazo robótico. 

El tubo de aceleración y guiado mide $L_0 = 30\text{ m}$ en reposo en el laboratorio. El radioisótopo utilizado para el seguimiento del haz tiene una vida media en reposo de $\tau_0 = 2.0 \times 10^{-8}\text{ s}$.

El haz a través del tubo acelerador tiene una velocidad de $v = 0.60c$:

1. Calcule la velocidad total del haz de partículas respecto al laboratorio usando el modelo clásico de Galileo y el modelo relativista de Einstein. Determine el porcentaje de error que cometería el sistema de control guiado si utilizara el modelo galileano.
2. Calcule la vida media del radioisótopo $\tau$ medida por el observador en el laboratorio durante el tránsito a $v = 0.60c$.
3. Calcule la longitud $L$ del tubo acelerador medida por un observador que viaja junto con las partículas del haz.
4. Calcule la fracción de núcleos activos $N/N_0$ que sobreviven al tiempo de tránsito por el tubo ($t_{\text{tránsito}} = L_0/v$) según el modelo galileano (usando $\tau_0$) y según el modelo relativista (usando $\tau$). Determine el factor de discrepancia entre ambos modelos.

### Datos del problema
* $c = 3.0 \times 10^8\text{ m/s}$
* $v_{\text{robot}} = 0.10c$
* $v_{\text{haz}} = 0.85c$
* $v = 0.60c$ (velocidad de tránsito del isótopo en el laboratorio)
* $L_0 = 30\text{ m}$
* $\tau_0 = 2.0 \times 10^{-8}\text{ s}$

---

### 1. Composición de Velocidades

#### A) Modelo Galileano:
$$v_{\text{Galileo}} = v_{\text{haz}} + v_{\text{robot}}$$

$$v_{\text{Galileo}} = 0.85c + 0.10c = 0.95c$$

#### B) Modelo Relativista:
$$v_{\text{Einstein}} = \frac{v_{\text{haz}} + v_{\text{robot}}}{1 + \frac{v_{\text{haz}} \cdot v_{\text{robot}}}{c^2}}$$

$$v_{\text{Einstein}} = \frac{0.85c + 0.10c}{1 + \frac{(0.85c)(0.10c)}{c^2}} = \frac{0.95c}{1.085} \approx 0.8756c$$

$$v_{\text{Einstein}} \approx 0.876c$$

#### C) Porcentaje de sobreestimación del modelo galileano:
$$\text{Error (\%)} = \left( \frac{v_{\text{Galileo}} - v_{\text{Einstein}}}{v_{\text{Einstein}}} \right) \times 100$$

$$\text{Error (\%)} = \left( \frac{0.95c - 0.8756c}{0.8756c} \right) \times 100 \approx 8.50\%$$

**Interpretacion:**  Si el software de control calcula la trayectoria y el tiempo de vuelo de la partícula mediante la composición galileana, introducirá un desfase espacial en el punto de impacto, afectando el margen de tejido sano adyacente al tumor.

---

### 2. Dilatación Temporal

#### Factor de Lorentz ($\gamma$):
$$\gamma = \frac{1}{\sqrt{1 - \left(\frac{v}{c}\right)^2}} = \frac{1}{\sqrt{1 - (0.60)^2}} = \frac{1}{\sqrt{0.64}} = 1.25$$

#### Vida media dilatada ($\tau$):
$$\tau = \gamma \cdot \tau_0$$

$$\tau = 1.25 \times (2.0 \times 10^{-8}\text{ s})$$

$$\tau = 2.5 \times 10^{-8}\text{ s}$$

### 3. Contracción de Longitud

$$
L = \frac{L_0}{\gamma} = \frac{30\,\text{m}}{1.25} = 24\,\text{m}
$$

**Interpretación:** El observador en el laboratorio explica la mayor supervivencia de núcleos debido a la **dilatación del tiempo** ($\tau$ aumenta), mientras que el observador que viaja con el haz observa una **contracción de la longitud** del tubo ($24\text{ m}$ en lugar de $30\text{ m}$), necesitando menos tiempo propio para atravesarlo. Ambos marcos llegan a la misma conclusión: sobreviven más partículas porque la física relativista acorta la duración del viaje en relación con la vida del isótopo.

### 4. Análisis Dosimétrico

#### Tiempo de tránsito en el marco del laboratorio:
$$t_{\text{tránsito}} = \frac{L_0}{v} = \frac{30}{1.8 \times 10^8}\text{ s} \approx 1.667 \times 10^{-7}\text{ s}$$

#### A) Modelo Galileano (usa $\tau_0 = 2.0 \times 10^{-8}\text{ s}$):
$$\frac{N}{N_0} = e^{-\left(\frac{t_{\text{tránsito}}}{\tau_0}\right)} = e^{-8.333} \approx 0.00024 \quad (0.024\%)$$

#### B) Modelo Relativista (usa $\tau = 2.5 \times 10^{-8}\text{ s}$):
$$\frac{N}{N_0} = e^{-\left(\frac{t_{\text{tránsito}}}{\tau}\right)} = e^{-6.667} \approx 0.00127 \quad (0.127\%)$$

#### C) Factor de Discrepancia:
$$\text{Factor} = \frac{(N/N_0)_{\text{Relativista}}}{(N/N_0)_{\text{Galileano}}} = \frac{0.00127}{0.00024} \approx 5.29$$

El modelo relativista predice **$5.29$ veces más actividad radiológica** al final del trayecto que el modelo clásico.
Al asumir un tiempo absoluto ($\tau = \tau_0$), el modelo galileano asume que el trazador decae más rápido de lo que ocurre en realidad. Por lo tanto, **subestima la radiación activa que llega al detector** ($0.024\%$ predecido frente a $0.127\%$ real).
